#### Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

#### Load Datasets

```python
train = pd.read_csv("../data/raw/train.csv")

features = pd.read_csv("../data/raw/features.csv")

stores = pd.read_csv("../data/raw/stores.csv")

test = pd.read_csv("../data/raw/test.csv")
```

#### Check Data Types

```python
train.dtypes
```

```python
features.dtypes
```

```python
stores.dtypes
```

####  Convert Date Columns

```python
train["Date"] = pd.to_datetime(train["Date"])
features["Date"] = pd.to_datetime(features["Date"])
test["Date"] = pd.to_datetime(test["Date"])
```

#### Check Missing Values

```python
train.isnull().sum()
```

```python
features.isnull().sum()
```

```python
train.isnull().sum().sort_values(ascending=False)
```

#### Missing Value Percentage

```python
missing = (
    train.isnull()
         .sum()
         .to_frame("Missing")
)

missing["Percentage"] = (
    missing["Missing"] / len(train)
) * 100

missing.sort_values(
    "Percentage",
    ascending=False
)
```

#### Handling Missing value

```python
markdown_cols = [
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

features[markdown_cols] = features[markdown_cols].fillna(0)
```

```python
features.head()
```

```python
features["CPI"].isnull().sum()
```

```python
features["CPI"] = features["CPI"].fillna(
    features["CPI"].median()
)
```

```python
features.head()
```

```python
features["Unemployment"].isnull().sum()
```

```python
features["Unemployment"] = features["Unemployment"].fillna(
    features["Unemployment"].median()
)
```

```python
features.head()
```

#### Check Duplicate Records

```python
train.duplicated().sum()
```

```python
features.duplicated().sum()
```

```python
stores.duplicated().sum()
```

#### Negative Weekly Sales

```python
(train["Weekly_Sales"] < 0).sum()
```

```python
train.loc[
    train["Weekly_Sales"] < 0,
    ["Store", "Dept", "Date", "Weekly_Sales"]
].head()
```

#### Validate Store Types

```python
train["Type"].value_counts()
```

#### Validate Holiday Values

```python
train["IsHoliday"].value_counts()
```

#### Merge Datasets

```python
train = train.merge(
    stores,
    on="Store",
    how="left"
)
```

```python
train = train.merge(
    features,
    on=["Store", "Date", "IsHoliday"],
    how="left"
)
```

#### Verify Merge Results

```python
train.shape
```

```python
train.head()
```

```python

#Check for missing store information.
train["Type"].isnull().sum()
```

```python
#Check missing feature values.
train["Temperature"].isnull().sum()
```

#### Final Missing Value Check

```python
missing = (
    train.isnull()
         .sum()
         .sort_values(ascending=False)
)

missing
```

#### Save the Cleaned Dataset

```python
train.to_csv(
    "../data/processed/train_clean.csv",
    index=False
)
```

### Key Observations

- Date columns were converted to the correct datetime format.
- Promotional MarkDown columns contained a high proportion of missing values, which were interpreted as weeks without promotions and replaced with 0.
- Missing values in CPI and Unemployment were imputed using the median to preserve the overall distribution.
- No duplicate records were found after cleaning.
- Negative weekly sales were retained because they likely represent product returns or financial adjustments.
- Sales, store, and external feature datasets were successfully merged into a single analytical table.
- The final dataset contains complete store information and external factors required for forecasting.
