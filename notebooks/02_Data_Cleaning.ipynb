# 02 - Data Cleaning

## Business Objective

The objective of this notebook is to clean and prepare the Walmart Sales Forecasting datasets for exploratory data analysis and predictive modeling. This involves identifying and handling data quality issues, converting data types, treating missing values, validating dataset relationships, and creating a consistent analytical dataset that can be used for feature engineering and model development.

## Analytical Objectives

- Convert date columns to the **datetime** data type.
- Check and confirm missing values in all datasets.
- Handle missing values using appropriate imputation techniques.
- Verify that no duplicate records exist.
- Validate data types for all columns.
- Check for invalid or inconsistent values.
- Investigate negative `Weekly_Sales` values.
- Merge the **train**, **features**, and **stores** datasets.
- Merge the **test**, **features**, and **stores** datasets.
- Verify that all joins preserve the expected number of records.
- Confirm that the cleaned datasets contain no unexpected missing values.
- Save the cleaned datasets for use in Exploratory Data Analysis (EDA) and machine learning.

## Expected Outcome

By the end of this notebook:

- All datasets will have consistent and appropriate data types.
- Missing values will be handled using suitable preprocessing techniques.
- The training and testing datasets will be successfully merged with the external features and store information.
- The cleaned datasets will be validated for quality and consistency.
- The final analytical datasets will be ready for exploratory data analysis, feature engineering, and forecasting model development.
